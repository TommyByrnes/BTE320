```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)


class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4
        self.width = self.getWidth()
        self.height = self.getHeight()

    def getWidth(self):
        return self.p1.distance(self.p2)

    def getHeight(self):
        return self.p2.distance(self.p3)

    def getArea(self):
        return self.width * self.height


p1 = Point(0, 0)
p2 = Point(4, 0)
p3 = Point(4, 3)
p4 = Point(0, 3)
r = Rectangle(p1, p2, p3, p4)
print(f"Width: {r.getWidth()}, Height: {r.getHeight()}, Area: {r.getArea()}")
```
