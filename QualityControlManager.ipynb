```python
def getValues(n):
    values = []
    for i in range(n):
        while True:
            try:
                val = float(input(f"Enter value {i+1}: "))
                if val < 0:
                    raise ValueError
                values.append(val)
                break
            except ValueError:
                print("Invalid input. Enter a positive number.")
    return values

def calcAverage(alist):
    trimmed = alist[3:]
    return sum(trimmed) / len(trimmed)

n = int(input("How many values? "))
values = getValues(n)
print(f"Average (excluding first 3): {calcAverage(values):.2f}")
```
