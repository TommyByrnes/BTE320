```python
def loan_payment(principal, annual_rate, years):
    if principal <= 0 or annual_rate <= 0 or years <= 0:
        raise ValueError("All inputs must be positive numbers")

    r = annual_rate / 12
    n = years * 12
    payment = (r * principal) / (1 - (1 + r) ** -n)
    return payment

print(loan_payment(200000, 0.06, 30))
```
