```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__wage = wage
        self.hours = hours

    def get_wage(self):
        return self.__wage

    def set_wage(self, wage):
        if wage < 0:
            print("Wage cannot be negative.")
        else:
            self.__wage = wage

    def calculate_pay(self):
        return self.__wage * self.hours

    def display_info(self):
        print(f"Name: {self.name}, Wage: ${self.__wage}/hr, Hours: {self.hours}, Pay: ${self.calculate_pay():.2f}")


class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):
        return super().calculate_pay() + self.bonus

    def display_info(self):
        print(f"Name: {self.name}, Wage: ${self.get_wage()}/hr, Hours: {self.hours}, Bonus: ${self.bonus}, Pay: ${self.calculate_pay():.2f}")


e = Employee("Tommy", 20, 40)
m = Manager("Louis", 35, 40, 500)

e.display_info()
m.display_info()
```
