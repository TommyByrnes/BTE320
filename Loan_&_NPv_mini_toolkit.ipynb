```python
def roi(cost, revenue):
    assert isinstance(cost, (int, float))
    assert isinstance(revenue, (int, float))
    assert cost != 0
    return (revenue - cost) / cost

def loan_payment(principal, annual_rate, months):
    assert isinstance(principal, (int, float))
    assert isinstance(annual_rate, (int, float))
    assert isinstance(months, (int, float))
    if principal <= 0:
        raise ValueError("principal must be positive")
    if months <= 0:
        raise ValueError("months must be positive")
    if annual_rate < 0:
        raise ValueError("annual_rate cannot be negative")
    r = annual_rate / 12
    if r != 0:
        pmt = r * principal / (1 - (1 + r) ** (-months))
    else:
        pmt = principal / months
    return pmt


try:
    cost = float(input("Enter cost: "))
    revenue = float(input("Enter revenue: "))
    print(roi(cost, revenue))
except Exception as e:
    print(e)


try:
    principal = float(input("Enter principal: "))
    annual_rate = float(input("Enter annual rate (decimal): "))
    months = int(input("Enter months: "))
    print(round(loan_payment(principal, annual_rate, months), 2))
except Exception as e:
    print(e)
```
