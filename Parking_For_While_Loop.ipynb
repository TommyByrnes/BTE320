```python
for hours in range(1, 9):
    price = 5 + 2.50 * hours
    if price < 10:
        price = 10
    if price > 20:
        price = 20
    print(hours, price)
```

```python
hours = 1
while hours < 9:
    price = 5 + 2.50 * hours

    if price < 10:
        price = 10
    if price > 20:
        price = 20

    print(hours, price)

    hours += 1
```
