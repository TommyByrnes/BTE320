```python
import math

ticket_price = 10
base_attendance = 20
fixed_cost = 200

print("Ads\tProfit")

for ads in range(0, 201, 25):
    additional_attendees = 2 * math.sqrt(ads)
    total_attendees = base_attendance + additional_attendees

    revenue = total_attendees * ticket_price
    profit = revenue - fixed_cost - ads

    print(f"{ads:.1f}\t{profit:.1f}")
```

```python
import math

ticket_price = 10
base_attendance = 20
fixed_cost = 200

print("Ads\tProfit")

ads = 0
while ads <= 200:
    additional_attendees = 2 * math.sqrt(ads)
    total_attendees = base_attendance + additional_attendees

    revenue = total_attendees * ticket_price
    profit = revenue - fixed_cost - ads

    print(f"{ads:.1f}\t{profit:.1f}")

    ads += 25
```
