```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return f"{self.first} {self.last}"

p = Greeting("Tommy", "Smith")
p.display(True)
p.display(False)
print(p)
```
